Plain text
## Conclusão

A análise mostrou que Ed Sheeran possui o maior número de visualizações acumuladas no YouTube dentro deste conjunto de dados. Outros artistas como CoComelon, Katy Perry, Justin Bieber e Coldplay também aparecem entre os mais visualizados.


```python
import matplotlib.pyplot as plt

top_artistas = df.groupby('Artist')['Views'].sum().sort_values(ascending=False).head(10)

top_artistas.plot(kind='bar')

plt.title('Top 10 artistas com mais visualizações')
plt.xlabel('Artista')
plt.ylabel('Visualizações')
plt.show()
```
